## Worked example

`CountVectorizer` turns each tweet into word counts; `LogisticRegression` learns which words point to positive or negative.
`make_pipeline` joins them, so the model takes raw text in.

```python
import pandas as pd
from sklearn.pipeline import make_pipeline
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.dummy import DummyClassifier
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split

tweets = pd.read_csv("data/naijasenti_pcm_en.csv")
print(tweets.groupby(["variety", "label"]).size())

en = tweets[tweets["variety"] == "english_like"]
pcm = tweets[tweets["variety"] == "pidgin"]
train_en, test_en = train_test_split(en, test_size=0.25, random_state=42, stratify=en["label"])
train_pcm, test_pcm = train_test_split(pcm, test_size=0.25, random_state=42, stratify=pcm["label"])
print(len(train_en), len(test_en), len(train_pcm), len(test_pcm))
```

```python
# A tiny example
demo = make_pipeline(CountVectorizer(), LogisticRegression())
demo.fit(["I love this", "this is great", "I hate this", "this is awful"], ["positive", "positive", "negative", "negative"])
demo.predict(["great stuff", "awful stuff"])
```
